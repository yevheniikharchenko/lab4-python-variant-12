```python
def get_flight_notification(h:float, v_y:float) -> str:
    match (h, v_y):
        case (h, v_y) if h < 300.0 and v_y > 15.0:
            return "PULL UP! (Негайний набір висоти)"
        case (h, v_y) if h < 100.0:
            return "TOO LOW! (Низька висота)"
        case _:
            return "Параметри безпечні"

print(get_flight_notification(150, 16))
print(get_flight_notification(99, 15))
print(get_flight_notification(150, 15))
```

```python
def id_flight_state(state:dict) -> str:
    match state:
        case {"power" : p, "thrust" : t, "mode" : m} if m == "BOOST" and p > 5000:
            return "Currently boosting"
        case {"power" : p, "thrust" : t, "mode" : m} if m == "ECO":
            return "Currently eco"
        case {"power" : p, "thrust" : t, "mode" : m}:
            return f"Unid-ed state: {m}"
        case _:
            return "Invalid input"

print(id_flight_state({"power" : 6000, "thrust" : 200, "mode" : "BOOST"}))
print(id_flight_state({"power" : 500, "thrust" : 1, "mode" : "ECO"}))
print(id_flight_state({"power" : 3000, "thrust" : 100, "mode" : "D"}))
print(id_flight_state("jhwevskjvs"))
```

```python
def RAM_check(percentages: list) -> bool:
    for p in percentages:
        if p > 0.95:
            print("Скидання буфера на Flash-пам'ять")
            return True
    return False

print(RAM_check([.15,.96,.90,.50]))
print(RAM_check([.80,.07,.90,.10]))
print(RAM_check([.03,.92,.90,.50]))
print(RAM_check([1.0,.96,.90,.50]))
```

```python
def get_TPU_rising(T : list, P : list) -> list:
    match T,P:
        case [*_],[*_] if len(T) == len(P):
            result = []
            for index, (t, p) in enumerate(zip(T, P)):
                if (t > 0) and (p > 0):
                    result.append(index)
            return result
        case [*_],[*_]:
            print("Списки похідних мають різну довжину!")
            return []
        case _:
            print("Некоректні дані!")
            return []

print(get_TPU_rising([1,1,1],[2,2,2]))
print(get_TPU_rising([1,1,1],[2,2]))
print(get_TPU_rising("rokjgjnwrok", "fikwsjvokr"))
```
